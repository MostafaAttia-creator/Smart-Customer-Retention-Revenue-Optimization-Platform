# Data Cleaning & Preprocessing

## Objective

The objective of this notebook is to clean the dataset and prepare it for exploratory data analysis and machine learning. This includes handling incorrect data types, detecting hidden missing values, removing unnecessary features, and preventing data leakage.

# Import Libraries

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:.2f}".format)
```

# Load the Dataset

```python
df = pd.read_excel(r"..\data\raw_data.xlsx")

df.head()
```

# Create a Copy of the Dataset

```python
cleaned_df = df.copy()
```

# Dataset Information

```python
cleaned_df.info()
```

# Missing Values

```python
(cleaned_df == " ").sum().sort_values(ascending=False)
```

## Replace Blank Spaces with Missing Values

```python
cleaned_df = cleaned_df.replace(" ", np.nan)
```

## Missing Values After Replacement

```python
cleaned_df.isnull().sum().sort_values(ascending=False)
```

# Data Types

```python
cleaned_df.dtypes
```

# Data Types

```python
cleaned_df.dtypes
```

## Convert Total Charges to Numeric

```python
cleaned_df["Total Charges"] = pd.to_numeric(
    cleaned_df["Total Charges"],
    errors="coerce"
)
```

```python
cleaned_df["Total Charges"].dtype
```

# Missing Values Summary

```python
missing = pd.DataFrame({
    "Missing": cleaned_df.isnull().sum(),
    "Percentage": cleaned_df.isnull().mean() * 100
})

missing[missing["Missing"] > 0]
```

```python
cleaned_df.loc[
    cleaned_df["Total Charges"].isnull(),
    ["Tenure Months", "Monthly Charges", "Total Charges"]
]
```

```python
cleaned_df = cleaned_df.dropna(subset=["Total Charges"])
```

```python
cleaned_df.isnull().sum().sum()
```

```python
missing = cleaned_df.isnull().sum()

missing[missing > 0]
```

## Missing Values Analysis

The dataset contains missing values only in the **Churn Reason** feature.

This is expected because the reason for churn is only available for customers who have already left the company. Customers who did not churn do not have a churn reason.

Therefore, these missing values do not represent data quality issues.
